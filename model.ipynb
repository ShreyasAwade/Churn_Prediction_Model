Feature Engineering

```python
df.columns
```

```python
df.head(1)
```

```python
y = df[["Churn"]]
x = df[["Age","Gender","Tenure","MonthlyCharges"]]
```

```python
x
```

```python
y
```

Setting the Numeric Value for Gender 1--> Female 0 -->Male on x and 1--> Yes 0--> No on y (churn)

```python
x["Gender"] = x["Gender"].apply(lambda x:1 if x=="Female" else 0)
```

```python
y["Churn"] = y["Churn"] .apply(lambda x:1 if x=="Yes" else 0)
```

```python
y
```

Training and Testing Model

x → Input features (independent variables)
y → Output/target (dependent variable)
0.2 = 20% of the data
80% → Training data
20% → Testing data

Train Model --> x_train , y_train
Make Predictions --> x_test , y_test

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2)
```

```python
scaler = StandardScaler()
```

```python
x_train = scaler.fit_transform(x_train)
```

```python
joblib.dump(scaler,"scaler.pkl")
```

```python
x_test = scaler.fit_transform(x_test)
```

Classification Model

```python
def modelperformance(predictions):
    print("Accuracy Score for the Model is: {}".format(accuracy_score(y_test,predictions)))
```

Logistic Regression Model

```python
log_model = LogisticRegression()
```

```python
log_model.fit(x_train , y_train)
```

```python
y_pred = log_model.predict(x_test)
```

```python
modelperformance(y_pred)
```

K Neighbors

```python
param_grid = {
    "n_neighbors": [3,5,7,9],
    "weights" : ["uniform","distance"],
}
```

```python
gridkn = GridSearchCV(KNeighborsClassifier(),param_grid, cv=5)
```

```python
gridkn.fit(x_train,y_train)
```

```python
gridkn.best_params_
```

```python
y_pred = gridkn.predict(x_test)
```

```python
modelperformance(y_pred)
```

Support Vector Machine

```python
svm = SVC()
```

```python
param_grid={
    "C" : [0.01,0.1,0.5,1],
    "kernel" : ["linear","rbf","poly"]
}
```

```python
gridsvc = GridSearchCV(svm,param_grid,cv = 5)
```

```python
gridsvc.fit(x_train,y_train)
```

```python
gridsvc.best_params_
```

```python
y_pred = gridsvc.predict(x_test)
```

```python
modelperformance(y_pred)
```

Decision Tree Classifier

```python
param_grid = {
    "criterion" : ["gini","entropy"],
    "splitter" : ["best","random"],
    "max_depth" : [None,10,20,30],
    "min_samples_split": [2,5,10],
    "min_samples_leaf":[1,2,4]
}
```

```python
grid_tree = GridSearchCV(DecisionTreeClassifier(),param_grid,cv=5)
```

```python
grid_tree.fit(x_train,y_train)
```

```python
grid_tree.best_params_
```

```python
y_pred = grid_tree.predict(x_test)
```

```python
y_pred
```

```python
modelperformance(y_pred)
```
